### Writing a function performing TEXT cleaning before vectorization 

```python
import pandas as pd 
import re
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

def text_cleaning(text):
    text = text.lower()
    # lowercase 

    text = re.sub('[^a-zA-Z\s]','',text)
    #removing special characters

    words = text.split()
    #tokenization

    words = [word for word in words if word not in stop_words]
    # removing stopwords

    words = [lemmatizer.lemmatize(word) for word in words]
    # lemmatization 

    return ' '.join(words)
```

```python
df['cleaned_text'] = df['text'].apply(text_cleaning) # applying cleaning function to features 
```

## Step 1- Splitting inputs (X) and output (y)

```python
# indentifying our target variable and splitting Inputs(features) X and Output y
```

```python
X = df['cleaned_text']
y = df['category']
```

## Step 2- Splitting the data into TEST(X_train, y_train) and TRAIN(X_test)

```python
# As we cannot pass the actual/whole data directly to the model we are Splitting TEST(X_train, y_train) and TRAIN(X_test) data with 80-20 ratio
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,train_size = 0.8,random_state=45)

```

## Step 3 - Vectorization 

#### Transforming corpus to DTM (Document-Term Matrix)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

TFIDF = TfidfVectorizer(max_features=5000)
X_train_trans = TFIDF.fit_transform(X_train) 
```

```python
X_train_trans.shape
```

## Step 3.5 - Feature Engineering on Target y_train

```python
# as y_train is in categorical form we are encoding it using Label encoding
```

```python
from sklearn.preprocessing import LabelEncoder

LE = LabelEncoder()

y_train_trans = LE.fit_transform(y_train)
```

```python
y_train_trans
```

```python
y_train_trans
```

## Step 4- Model training using Algorithmn X_test_trans and y_train

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier()
model.fit(X_train_trans,y_train_trans)
```

## Step 5 - Vectorization and Feature engineering on Test data 

```python
X_test_trans = TFIDF.transform(X_test)
```

```python
X_test_trans
```

```python
y_test_trans = LE.transform(y_test)
```

```python
y_test_trans
```

## Step 6 - Prediction on X_test_trans

#### here we are pasing our test inputs (X_test_trans) to the model to get our prediction output(y_pred) 

```python
y_pred =  model.predict(X_test_trans)
```

```python
y_pred
```

## Step 7 Passing y-test(actual) and y_pred(model_prediction) to evaluation metrics

#### in this step we are passing our actual target variable and model prediction to evaluation metrics 
- To check how accurate our model is performing

```python
from sklearn.metrics import accuracy_score
score = accuracy_score(y_test_trans,y_pred)
```
